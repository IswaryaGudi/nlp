```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
text="""Artificial Intelligence (AI) is a branch of computer science that focuses on developing intelligent machines and computer systems that can perform tasks that normally require human intelligence,
such as learning, reasoning, problem-solving, decision-making, understanding natural language, recognizing images, and speech recognition. The main goal of AI is to make machines capable of thinking,
understanding, and responding intelligently to real-world situations.
AI is widely used in applications such as virtual assistants, self-driving cars, chatbots, recommendation systems, healthcare, robotics, and face recognition.
Machine Learning is a subset of Artificial Intelligence that enables computers to learn automatically from data and improve their performance without being explicitly programmed for every task. Instead of providing fixed instructions for every situation, we train machine learning models using historical data so that they can identify patterns, understand relationships, make predictions, and take decisions based on new or unseen data. Machine Learning mainly consists of three types: Supervised Learning, Unsupervised Learning, and Reinforcement Learning. Supervised Learning uses labelled data to perform classification and regression, Unsupervised Learning identifies hidden patterns and groups in unlabelled data using techniques such as clustering and dimensionality reduction, and Reinforcement Learning allows an agent to learn through interaction with an environment using rewards and penalties. ML is widely used in spam detection, fraud detection, house price prediction, customer segmentation, recommendation systems, and medical diagnosis.
Artificial Intelligence is the broader concept of making machines intelligent, whereas Machine Learning is one of the approaches used to achieve artificial intelligence. Deep Learning (DL) is a further subset of Machine Learning that uses artificial neural networks with multiple layers to learn complex patterns from large amounts of data. In simple terms, AI is the overall field, ML is a subset of AI, and Deep Learning is a subset of ML. Together, these technologies play an important role in modern applications such as computer vision, natural language processing, autonomous vehicles, robotics, and generative AI."""
```

```python
#convert words into nbrs
tokenizer=Tokenizer()
tokenizer.fit_on_texts([text])
```

```python
#get total no of unique words
total_words=len(tokenizer.word_index)+1
print(total_words)
```

```python
#generate sequences for next word prediction
sequences=[]
for line in text.split('\n'):
    tokens=tokenizer.texts_to_sequences([line])[0]
    for i in range(1,len(tokens)):
        sequences.append(tokens[:i+1])
```

```python
sequences
```

```python
#find maximum sequence length
max_len=max(len(seq) for seq in sequences)
print(max_len)
```

```python
#make all sequences equal in length
sequences=pad_sequences(sequences,maxlen=max_len,padding='pre')
```

```python
sequences
```

```python
# x contains previous word and y contains the next word
x=sequences[:,:-1]
y=sequences[:,-1]
```

```python
#create LSTM model
model=Sequential()
#convert word numbers into vectors of 10 values
model.add(Embedding(total_words,10))
#learn patterns from previous words
model.add(LSTM(50))
#predict the next word from the vocabulary
model.add(Dense(total_words,activation='softmax'))
```

```python
#configure the model
model.compile(loss='sparse_categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
#train the model
model.fit(x,y,epochs=500)
print("model training completed")
```

```python
#starting sentence
seed_text="ml and ai"
next_words=20
for _ in range(next_words):
  #convert sentence into numbers
  token=tokenizer.texts_to_sequences([seed_text])[0]
  #apply padding
  tokens=pad_sequences([token],maxlen=max_len-1,padding='pre')
  #predict the next word
  prediction=model.predict(tokens)
  #select the word with highest probability
  predicted_index=np.argmax(prediction)
  #get the word from index
  next_word=tokenizer.index_word.get(predicted_index, '')
  #add the predicted word to the sentence
  seed_text=seed_text+" "+next_word
print(seed_text)
```
